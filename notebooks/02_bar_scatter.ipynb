# 🧩 Notebook-02: Bar and Scatter Plots with Matplotlib

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import sys
from pathlib import Path

# ✅ Make the project root importable
PROJECT_ROOT = Path.cwd().parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))  # makes the `mplmasterpro` package importable

from mplmasterpro.plot_utils import bar_plot, grouped_bar_plot, scatter_plot

print("📊 Bar and Scatter Plots using Matplotlib\n")
```

```python
# 🗂️ 1. Load Dataset
df = pd.read_csv(PROJECT_ROOT / "datasets" / "sales_data.csv")
df.head()
```

```python
# 📦 2. Vertical Bar Plot – Total Units Sold per Product
product_totals = df.groupby("Product")["Units Sold"].sum().reset_index()

bar_plot(
    categories=product_totals["Product"],
    values=product_totals["Units Sold"],
    title="Total Units Sold per Product",
    xlabel="Product",
    ylabel="Units Sold"
);
```

```python
# 📊 3. Horizontal Bar Plot – Revenue per Product
revenue_totals = df.groupby("Product")["Revenue"].sum().reset_index()

bar_plot(
    categories=revenue_totals["Product"],
    values=revenue_totals["Revenue"],
    horizontal=True,
    title="Total Revenue per Product",
    xlabel="Revenue ($)",
    ylabel="Product",
    color="salmon"
);
```

```python
# 🪄 4. Grouped Bar Plot – Product-Wise Sales Over Months
df_grouped = df.groupby(["Month", "Product"])["Units Sold"].sum().unstack().reset_index()
data_dict = {
    product: df_grouped[product].tolist() for product in df_grouped.columns[1:]
}

grouped_bar_plot(
    x_labels=df_grouped["Month"].tolist(),
    data_dict=data_dict,
    title="Monthly Units Sold by Product",
    xlabel="Month",
    ylabel="Units Sold"
);
```

```python
# 🎯 5. Scatter Plot – Units Sold vs Revenue
scatter_plot(
    x=df["Units Sold"],
    y=df["Revenue"],
    title="Revenue vs. Units Sold",
    xlabel="Units Sold",
    ylabel="Revenue"
);
```

```python
# 🎨 6. Advanced Scatter – Color + Size Mapping
df["ColorMap"] = df["Units Sold"]
df["BubbleSize"] = np.sqrt(df["Revenue"])

scatter_plot(
    x=df["Units Sold"],
    y=df["Revenue"],
    color_values=df["ColorMap"],
    size=df["BubbleSize"],
    cmap="viridis",
    title="Revenue vs. Units Sold (Colored & Scaled)",
    xlabel="Units Sold",
    ylabel="Revenue",
    use_colorbar=True,
    edgecolor="black"
);
```

## ✅ Summary
- ✅ Used reusable `bar_plot()`, `grouped_bar_plot()`, and `scatter_plot()` functions
- ✅ Built vertical, horizontal, and grouped bars
- ✅ Created bubble scatter plots with size & colormap
