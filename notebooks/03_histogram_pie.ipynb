# 🧩 Notebook-03: Histograms and Pie Charts with Matplotlib

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import sys
from pathlib import Path

# ✅ Make the project root importable
PROJECT_ROOT = Path.cwd().parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))  # makes the `mplmasterpro` package importable

from mplmasterpro.plot_utils import histogram_plot, pie_chart

print("📊 Histograms and Pie Charts using Matplotlib\n")
```

```python
# 🗂️ 1. Load Dataset
df = pd.read_csv(PROJECT_ROOT / "datasets" / "sales_data.csv")
df.head()
```

```python
# 📈 2. Histogram – Distribution of Units Sold
histogram_plot(
    data=df["Units Sold"],
    bins=10,
    color="skyblue",
    edgecolor="black",
    title="Distribution of Units Sold",
    xlabel="Units Sold",
    ylabel="Frequency"
);
```

```python
# 📊 3. Histogram – Revenue Distribution with Density
histogram_plot(
    data=df["Revenue"],
    bins=15,
    color="orange",
    edgecolor="white",
    alpha=0.8,
    density=True,
    title="Revenue Distribution (Density)",
    xlabel="Revenue"
);
```

```python
# 🥧 4. Pie Chart – Product Sales Proportion
product_totals = df.groupby("Product")["Units Sold"].sum().reset_index()

pie_chart(
    labels=product_totals["Product"],
    values=product_totals["Units Sold"],
    colors=["#66b3ff", "#99ff99", "#ffcc99", "#ff9999"],
    explode=[0.05]*4,
    title="Units Sold Share by Product"
);
```

```python
# 💰 5. Pie Chart – Revenue Proportion by Product
revenue_totals = df.groupby("Product")["Revenue"].sum().reset_index()

pie_chart(
    labels=revenue_totals["Product"],
    values=revenue_totals["Revenue"],
    colors=["#8ecae6", "#219ebc", "#ffb703", "#fb8500"],
    explode=[0.05]*4,
    title="Revenue Share by Product"
);
```

## ✅ Summary
- ✅ Used `histogram_plot()` for frequency and density plots
- ✅ Visualized revenue and unit distribution with bins
- ✅ Created styled pie charts for proportion analysis
- ✅ Introduced explode, custom colors, and shadows
