# 🧩 Notebook-05: Customization and Styling with Matplotlib

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import sys
from pathlib import Path

# ✅ Make the project root importable
PROJECT_ROOT = Path.cwd().parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))  # makes the `mplmasterpro` package importable

from mplmasterpro.plot_utils import line_plot
from mplmasterpro.theme_utils import apply_dark_theme, apply_corporate_theme, apply_minimal_theme, reset_theme

print("🎨 Customization and Styling in Matplotlib\n")
```

```python
# 🗂️ 1. Load Dataset

df = pd.read_csv(PROJECT_ROOT / "datasets" / "sales_data.csv")
monthly = df.groupby("Month")["Units Sold"].sum().reset_index()
```

```python
# 🎨 2. Default Theme (Baseline)

reset_theme()

line_plot(
    x=monthly["Month"],
    y=monthly["Units Sold"],
    title="Default Theme: Units Sold",
    xlabel="Month",
    ylabel="Units Sold",
    label="Units Sold",
    color="mediumblue",
    marker="o"
);
```

```python
# 🖤 3. Dark Theme
apply_dark_theme()

line_plot(
    x=monthly["Month"],
    y=monthly["Units Sold"],
    title="Dark Theme: Units Sold",
    xlabel="Month",
    ylabel="Units Sold",
    label="Units Sold",
    color="cyan",
    marker="D"
);
```

```python
# 🧑‍💼 4. Corporate Theme
apply_corporate_theme()

line_plot(
    x=monthly["Month"],
    y=monthly["Units Sold"],
    title="Corporate Theme: Units Sold",
    xlabel="Month",
    ylabel="Units Sold",
    label="Units Sold",
    color="#0077b6",
    marker="s"
);
```

```python
# ✨ 5. Minimal Theme
apply_minimal_theme()

line_plot(
    x=monthly["Month"],
    y=monthly["Units Sold"],
    title="Minimal Theme: Units Sold",
    xlabel="Month",
    ylabel="Units Sold",
    label="Units Sold",
    color="indigo",
    marker="^"
);
```

## ✅ Summary
- ✅ Applied 4 custom themes using rcParams: dark, corporate, minimal, and reset
- ✅ Encapsulated themes inside `theme_utils.py`
- ✅ Custom styling improves readability, aesthetics, and presentation
