# 🧩 Notebook-06: Advanced Plots with Matplotlib

```python
# 🧩 Notebook-06: Advanced Plots with Matplotlib

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import sys
from pathlib import Path

# ✅ Make the project root importable
PROJECT_ROOT = Path.cwd().parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))  # makes the `mplmasterpro` package importable

from mplmasterpro.plot_utils import fill_between_plot, log_scale_plot, twin_axes_fill_plot

print("🔬 Advanced Plots using Matplotlib\n")
```

```python
# 🗂️ 1. Load Dataset
df_sales = pd.read_csv(PROJECT_ROOT / "datasets" / "sales_data.csv")
df_stocks = pd.read_csv(PROJECT_ROOT / "datasets" / "stock_prices.csv")
```

```python
# 🌈 2. Fill Between – Units Sold Over Time
monthly_sales = df_sales.groupby("Month")["Units Sold"].sum().reset_index()

fill_between_plot(
    x=monthly_sales["Month"],
    y1=monthly_sales["Units Sold"],
    title="Area Under Total Units Sold",
    xlabel="Month",
    ylabel="Units Sold",
    color="lightcoral",
    label="Units Sold"
);
```

```python
# 🔬 3. Log Scale – Revenue Distribution (Log Y-axis)
revenue_by_order = df_sales["Revenue"].sort_values().reset_index(drop=True)

log_scale_plot(
    x=range(len(revenue_by_order)),
    y=revenue_by_order,
    log_axis="y",
    title="Revenue Distribution (Log-Scaled)",
    xlabel="Transactions",
    ylabel="Revenue ($)",
    color="darkorange",
    label="Revenue"
);
```

```python
# 📊 4. Log-Log Plot – Stock High vs Volume
df_tesla = df_stocks[df_stocks["Stock"] == "TSLA"]

log_scale_plot(
    x=df_tesla["Volume"],
    y=df_tesla["High"],
    log_axis="both",
    title="TSLA: High Price vs Volume (Log-Log)",
    xlabel="Volume",
    ylabel="High Price",
    color="green",
    marker="o",
    label="TSLA"
);
```

```python
# 🪄 5. Twin Axis Fill – Units Sold and Revenue
monthly_agg = df_sales.groupby("Month")[["Units Sold", "Revenue"]].sum().reset_index()

twin_axes_fill_plot(
    x=monthly_agg["Month"],
    y1=monthly_agg["Units Sold"],
    y2=monthly_agg["Revenue"],
    label1="Units Sold",
    label2="Revenue",
    xlabel="Month",
    ylabel1="Units Sold",
    ylabel2="Revenue ($)",
    title="Area Plot: Units Sold vs Revenue",
    color1="teal",
    color2="darkviolet"
);
```

# ✅ Summary
- ✅ Used `fill_between_plot()` to visualize trends over time
- ✅ Applied log scale using `log_scale_plot()` for better distribution views
- ✅ Created a log-log plot for stock data
- ✅ Used `twin_axes_fill_plot()` to overlay two scaled trends visually
