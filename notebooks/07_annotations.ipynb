# 🧩 Notebook-07: Annotations and Highlights with Matplotlib

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import sys
from pathlib import Path

# ✅ Make the project root importable
PROJECT_ROOT = Path.cwd().parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))  # makes the `mplmasterpro` package importable

from mplmasterpro.plot_utils import annotate_point, highlight_region, label_line

print("🖍️ Annotations and Highlights in Matplotlib\n")
```

```python
# 🗂️ 1. Load Dataset
df = pd.read_csv(PROJECT_ROOT / "datasets" / "sales_data.csv")
monthly = df.groupby("Month")[["Units Sold", "Revenue"]].sum().reset_index()
```

```python
# 📍 2. Annotate a Specific Point – Highest Units Sold
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(monthly["Month"], monthly["Units Sold"], marker='o', color='teal')

# Annotate max point
max_idx = monthly["Units Sold"].idxmax()
x_max = monthly["Month"][max_idx]
y_max = monthly["Units Sold"][max_idx]

annotate_point(ax, x=x_max, y=y_max, text="Highest Sales", textcolor='teal')
ax.set_title("Annotate Peak Units Sold")
ax.set_xlabel("Month")
ax.set_ylabel("Units Sold")
ax.tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 🌫️ 3. Highlight a Region – Holiday Spike (e.g., Nov–Dec)
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(monthly["Month"], monthly["Units Sold"], color='darkorange', marker='s')
highlight_region(ax, x_start="2023-11", x_end="2023-12", color="gold", alpha=0.4, label="Holiday Spike")
ax.set_title("Highlight Holiday Sales Region")
ax.set_xlabel("Month")
ax.set_ylabel("Units Sold")
ax.legend()
ax.tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 🏷️ 4. Label Line Directly – Revenue vs Units Sold (with twin axes)
fig, ax1 = plt.subplots(figsize=(10, 5))
x = monthly["Month"]
y1 = monthly["Units Sold"]
y2 = monthly["Revenue"]

ax1.plot(x, y1, color="blue", marker="o")
label_line(ax1, x=list(x), y=list(y1), text="Units Sold", location="right", color="blue", offset=(0, 5))
ax1.set_ylabel("Units Sold", color="blue")
ax1.tick_params(axis='y', labelcolor="blue")

ax2 = ax1.twinx()
ax2.plot(x, y2, color="green", marker="^")
label_line(ax2, x=list(x), y=list(y2), text="Revenue", location="right", color="green", offset=(0, -10))
ax2.set_ylabel("Revenue", color="green")
ax2.tick_params(axis='y', labelcolor="green")

ax1.set_xlabel("Month")
ax1.set_title("Label Lines Directly (with Twin Axes)")
ax1.tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.show()

plt.show()
```

# ✅ Summary
- ✅ Used `annotate_point()` to highlight max points
- ✅ Applied `highlight_region()` to mark time intervals
- ✅ Labeled curves directly with `label_line()` for clarity
