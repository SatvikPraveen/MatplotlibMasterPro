# 🧩 Notebook-10: Exporting and Styling Plots in Matplotlib

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import sys
from pathlib import Path

# ✅ Make the project root importable
PROJECT_ROOT = Path.cwd().parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))  # makes the `mplmasterpro` package importable

from mplmasterpro.plot_utils import save_plot, set_global_style

print("🖼️ Exporting & Styling with Matplotlib\n")
```

```python
# 🎨 1. Set Global Style
set_global_style(style_name="seaborn-v0_8-whitegrid", font="DejaVu Sans", size=12)
```

```python
# 📊 2. Load Data
df = pd.read_csv(PROJECT_ROOT / "datasets" / "sales_data.csv")
monthly = df.groupby("Month")[["Units Sold", "Revenue"]].sum().reset_index()
```

```python
# 📈 3. Generate Plot with Style
fig, ax = plt.subplots(figsize=(10, 5))
x = monthly["Month"]
y = monthly["Revenue"]

ax.plot(x, y, marker="o", color="purple")
ax.set_title("Monthly Revenue", fontsize=14)
ax.set_xlabel("Month")
ax.set_ylabel("Revenue ($)")
ax.tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.show()
```

```python
# 💾 4. Save Plot in Multiple Formats
save_plot(fig, filename="monthly_revenue", folder=PROJECT_ROOT / "exports")
```

## ✅ Summary
- 🎨 Applied a clean `seaborn-whitegrid` global style
- 🔤 Customized font and size across all plots
- 💾 Saved the plot to PNG, PDF, and SVG formats using `save_plot()`
- 📁 All files exported neatly into the `exports/` folder
