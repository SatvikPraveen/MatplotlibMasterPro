# 🧩 Notebook-11: Animated Line Plots

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import sys
from pathlib import Path

# ✅ Make the project root importable
PROJECT_ROOT = Path.cwd().parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))  # makes the `mplmasterpro` package importable

from mplmasterpro.plot_utils import animate_line_plot

print("🎞️ Animated Line Plots using Matplotlib Animation")
```

```python
# 📂 1. Load and Prepare Data
df = pd.read_csv(PROJECT_ROOT / "datasets" / "sales_data.csv")
monthly = df.groupby("Month")[["Units Sold", "Revenue"]].sum().reset_index()

x = monthly["Month"]
y = monthly["Revenue"]
```

```python
# 📈 2. Create Animation – Monthly Revenue Growth
animate_line_plot(
    x=np.arange(len(x)),
    y=y.values,
    xlabel="Month Index",
    ylabel="Revenue",
    title="Revenue Growth Over Time (Animated)",
    interval=400,
    color="green"
);
```

```python
# 🔁 3. Animate Dual Line Plot
from mplmasterpro.plot_utils import animate_dual_line_plot

animate_dual_line_plot(
    x=monthly["Month"],
    y1=monthly["Units Sold"],
    y2=monthly["Revenue"],
    xlabel="Month",
    ylabel="Value",
    labels=("Units Sold", "Revenue"),
    title="Animate Units Sold and Revenue"
);
```

```python
# 📊 3. Animated Bar Chart
from mplmasterpro.plot_utils import animate_bar_chart

product = df.groupby("Product")["Revenue"].sum().sort_values()
animate_bar_chart(
    categories=product.index.tolist(),
    values=product.values.tolist(),
    xlabel="Product",
    ylabel="Revenue",
    title="Animated Product-wise Revenue"
);
```

```python
# 🔵 4. Animated Scatter Growth
from mplmasterpro.plot_utils import animate_scatter_growth

animate_scatter_growth(
    x=df["Units Sold"],
    y=df["Revenue"],
    xlabel="Units Sold",
    ylabel="Revenue",
    title="Revenue vs Units Sold Growth"
);
```

```python
# 📦 5. Export Animations as .mp4

from mplmasterpro.plot_utils import (
    save_line_animation,
    save_dual_line_animation,
    save_bar_animation,
    save_scatter_animation
)

# 🔃 Export: Single Line Plot
save_line_animation(
    x=np.arange(len(x)),
    y=y.values,
    filename="../exports/revenue_growth.mp4"
)

# 🔁 Export: Dual Line Plot
save_dual_line_animation(
    x=np.arange(len(monthly["Month"])),
    y1=monthly["Units Sold"].values,
    y2=monthly["Revenue"].values,
    labels=("Units Sold", "Revenue"),
    filename="../exports/units_revenue_growth.mp4"
)

# 📊 Export: Bar Chart Animation
save_bar_animation(
    categories=product.index.tolist(),
    values=product.values.tolist(),
    filename="../exports/product_revenue_bars.mp4"
)

# 🔵 Export: Scatter Animation
save_scatter_animation(
    x=df["Units Sold"].values,
    y=df["Revenue"].values,
    filename="../exports/revenue_vs_units_scatter.mp4"
);
```
