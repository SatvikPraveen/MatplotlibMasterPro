## 📍 1. Basic 3D Scatter Plot

```python
# Generate random 3D data
np.random.seed(42)
n = 100
x = np.random.randn(n)
y = np.random.randn(n)
z = np.random.randn(n)
colors = np.random.rand(n)
sizes = 1000 * np.random.rand(n)

# Create 3D scatter plot
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

scatter = ax.scatter(x, y, z, c=colors, s=sizes, alpha=0.6, cmap='viridis', edgecolors='black', linewidth=0.5)

ax.set_xlabel('X Axis', fontsize=12, fontweight='bold')
ax.set_ylabel('Y Axis', fontsize=12, fontweight='bold')
ax.set_zlabel('Z Axis', fontsize=12, fontweight='bold')
ax.set_title('3D Scatter Plot with Random Data', fontsize=14, fontweight='bold', pad=20)

# Add colorbar
cbar = plt.colorbar(scatter, ax=ax, shrink=0.5, aspect=5)
cbar.set_label('Value', rotation=270, labelpad=15)

plt.tight_layout()
plt.show()
```

## 📈 2. 3D Line Plot (Parametric Curve)

```python
# Create a parametric 3D spiral
theta = np.linspace(-4 * np.pi, 4 * np.pi, 100)
z = np.linspace(-2, 2, 100)
r = z**2 + 1
x = r * np.sin(theta)
y = r * np.cos(theta)

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

ax.plot(x, y, z, 'b-', linewidth=2, label='Parametric Spiral')
ax.scatter(x[0], y[0], z[0], c='green', s=100, marker='o', label='Start')
ax.scatter(x[-1], y[-1], z[-1], c='red', s=100, marker='s', label='End')

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('3D Parametric Curve (Spiral)', fontsize=14, fontweight='bold')
ax.legend()

plt.show()
```

## 🌊 3. 3D Surface Plot

```python
# Create meshgrid for surface plot
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

# Create surface plot
surf = ax.plot_surface(X, Y, Z, cmap='coolwarm', alpha=0.8, 
                       antialiased=True, edgecolor='none')

ax.set_xlabel('X', fontsize=11)
ax.set_ylabel('Y', fontsize=11)
ax.set_zlabel('Z', fontsize=11)
ax.set_title('3D Surface Plot: $z = \sin(\sqrt{x^2 + y^2})$', fontsize=14, fontweight='bold')

# Add colorbar
fig.colorbar(surf, shrink=0.5, aspect=5)

# Set viewing angle
ax.view_init(elev=30, azim=45)

plt.show()
```

## 🕸️ 4. Wireframe Plot

```python
# Create data for wireframe
x = np.linspace(-5, 5, 30)
y = np.linspace(-5, 5, 30)
X, Y = np.meshgrid(x, y)
Z = np.exp(-(X**2 + Y**2) / 10) * np.cos(X) * np.sin(Y)

fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

# Create wireframe
wire = ax.plot_wireframe(X, Y, Z, color='darkblue', linewidth=0.8, alpha=0.7)

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('3D Wireframe Plot', fontsize=14, fontweight='bold')

ax.view_init(elev=25, azim=120)

plt.show()
```

## 🎨 5. Contour3D Plot

```python
# Create data
x = np.linspace(-3, 3, 50)
y = np.linspace(-3, 3, 50)
X, Y = np.meshgrid(x, y)
Z = np.sin(X) * np.cos(Y)

fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

# Create 3D contour plot
contour = ax.contour3D(X, Y, Z, 50, cmap='plasma')

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('3D Contour Plot', fontsize=14, fontweight='bold')

fig.colorbar(contour, shrink=0.5, aspect=5)

plt.show()
```

## 🔄 6. Multiple 3D Plots with Different Views

```python
# Create sample data
x = np.linspace(-3, 3, 40)
y = np.linspace(-3, 3, 40)
X, Y = np.meshgrid(x, y)
Z = X**2 - Y**2

# Create figure with multiple 3D subplots
fig = plt.figure(figsize=(16, 6))

# View 1: Default angle
ax1 = fig.add_subplot(131, projection='3d')
ax1.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8)
ax1.set_title('View 1: Default (30°, -60°)', fontsize=11, fontweight='bold')
ax1.view_init(elev=30, azim=-60)

# View 2: Top-down
ax2 = fig.add_subplot(132, projection='3d')
ax2.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8)
ax2.set_title('View 2: Top-Down (90°, -90°)', fontsize=11, fontweight='bold')
ax2.view_init(elev=90, azim=-90)

# View 3: Side view
ax3 = fig.add_subplot(133, projection='3d')
ax3.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8)
ax3.set_title('View 3: Side View (0°, 0°)', fontsize=11, fontweight='bold')
ax3.view_init(elev=0, azim=0)

plt.tight_layout()
plt.show()
```

## 🎲 7. Real-World Example: 3D Scatter with Data Categories

```python
# Simulate 3D data for different categories
np.random.seed(42)

# Category A
x_a = np.random.randn(50) + 1
y_a = np.random.randn(50) + 1
z_a = np.random.randn(50) + 1

# Category B
x_b = np.random.randn(50) - 1
y_b = np.random.randn(50) + 1
z_b = np.random.randn(50) - 1

# Category C
x_c = np.random.randn(50)
y_c = np.random.randn(50) - 1
z_c = np.random.randn(50) + 2

fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot each category with different colors and markers
ax.scatter(x_a, y_a, z_a, c='red', marker='o', s=50, alpha=0.7, label='Category A')
ax.scatter(x_b, y_b, z_b, c='blue', marker='^', s=50, alpha=0.7, label='Category B')
ax.scatter(x_c, y_c, z_c, c='green', marker='s', s=50, alpha=0.7, label='Category C')

ax.set_xlabel('Feature 1', fontsize=11)
ax.set_ylabel('Feature 2', fontsize=11)
ax.set_zlabel('Feature 3', fontsize=11)
ax.set_title('3D Classification Visualization', fontsize=14, fontweight='bold', pad=20)
ax.legend(loc='upper left', fontsize=10)

ax.view_init(elev=20, azim=45)

plt.show()
```

## 📊 8. Bar Plot in 3D

```python
# Create 3D bar chart data
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

# Sample data for bars
_x = np.arange(4)
_y = np.arange(5)
_xx, _yy = np.meshgrid(_x, _y)
x, y = _xx.ravel(), _yy.ravel()

# Heights of bars
top = x + y
bottom = np.zeros_like(top)
width = depth = 0.5

# Create bars
colors = plt.cm.viridis(top / top.max())
ax.bar3d(x, y, bottom, width, depth, top, shade=True, color=colors, alpha=0.8)

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Value')
ax.set_title('3D Bar Chart', fontsize=14, fontweight='bold')

plt.show()
```

## 🎯 Key Takeaways

- ✅ Use `projection='3d'` to create 3D axes
- ✅ `scatter()`, `plot()`, `plot_surface()`, and `plot_wireframe()` work in 3D
- ✅ Use `view_init(elev, azim)` to set viewing angles
- ✅ Colormaps and colorbars work the same way as 2D plots
- ✅ 3D plots are great for visualizing multi-dimensional data and surfaces
- ✅ Wireframes are useful for understanding structure without occlusion
